# Raw flat type and flat model values

This notebook reads the supplied CSVs directly. It does not apply categorical cleaning or preprocessing.

```python
from pathlib import Path
import os
import pandas as pd

configured_root = os.environ.get('KDDM_PROJECT_ROOT')
candidate_roots = ([Path(configured_root)] if configured_root else []) + [Path.cwd(), *Path.cwd().parents]
project_root = next((path for path in candidate_roots if (path / 'data' / 'train.csv').exists()), None)
if project_root is None:
    raise FileNotFoundError('Set KDDM_PROJECT_ROOT to the project directory.')

raw_train = pd.read_csv(project_root / 'data' / 'train.csv', usecols=['FLAT_TYPE', 'FLAT_MODEL'])
raw_test = pd.read_csv(project_root / 'data' / 'test.csv', usecols=['FLAT_TYPE', 'FLAT_MODEL'])
```

```python
for dataset_name, frame in [('train', raw_train), ('test', raw_test)]:
    print(f'\n### Raw {dataset_name}.csv')
    for column in ['FLAT_TYPE', 'FLAT_MODEL']:
        counts = frame[column].value_counts(dropna=False)
        print(f'\n{column}: {len(counts)} unique values (including missing)')
        display(counts.rename_axis(column).reset_index(name='rows'))
        print('Unique values:', counts.index.tolist())
```
